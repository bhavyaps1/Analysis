### V. Exploratory Data Analysis and visualization

1. Which age groups, genders, occupations and countries are represented most?




##### Customer age groups

```python
bins=[0,18,25,35,50,100]
labels=['Under 18','18-25','26-35','36-50','50+']
df['Age_group']=pd.cut(df['Age'],bins=bins,labels=labels)
```

```python
df['Age_group'].value_counts().reset_index(name='Customer_count')
```

```python
sns.countplot(data=df,x='Age_group',order=labels)
plt.title('Customers by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

Insight: The 18–25 age group has the highest number of respondents, with 273 participants. This is followed by the 26–35 age group with 110 respondents. The Under 18 group has 33 respondents, while the 36–50 group has the lowest count, with 54 respondents.

Recommendation: Targeting younger consumers may benefit from focusing on digital platforms, mobile-friendly shopping experiences, and promotional strategies that appeal to this age group.

```python
gender_counts=df['Gender'].value_counts()
plt.pie(gender_counts.values,labels=gender_counts.index,autopct='%1.1f%%',startangle=90,wedgeprops={'width': 0.6,'edgecolor':'white'})
plt.title('Gender Distribution')
plt.show()
```

Insight: Male respondents form the largest group with 256 participants, accounting for approximately 52.9% of the dataset. Female respondents account for 177 participants(36.6%), while 51 respondents(10.5%) selected "Prefer not to say". The remaining 16 responses(3.2%) have missing gender information.

Recommendation: Here, most respondents are male. Consider collecting feedback from a more balanced range of customer groups to understand whether shopping preferences differ across gender categories.

```python
occupation_counts=df['Occupation_Status'].value_counts()
sns.barplot(x=occupation_counts.values,y=occupation_counts.index)
plt.title('Customers by Occupation Status')
plt.xlabel('Number of Customers')
plt.ylabel('Occupation Status')
plt.tight_layout()
plt.show()
```

Insight: Nearly half of the survey participants are students. This indicates that the dataset largely represents the shopping behaviour of students, followed by individuals working in full-time employment.

Recommendation: Consider student-focused discounts, budget-friendly product options, and promotional campaigns for working professionals to address the needs of the major groups represented in the survey.

```python
country_counts=df['Country_Region'].value_counts().head(10)
sns.barplot(x=country_counts.values,y=country_counts.index)
plt.title('Top 10 Countries / Regions by Survey Responses')
plt.xlabel('Number of Responses')
plt.ylabel('Country / Region')
plt.tight_layout()
plt.show()
```

3. Factors influencing the purchase decision? How important are reviews, social advertisements, discounts and price comparisons?

Comparison of marketing influence by customer reviews and social media ads:

```python
influence_comparison=df[['Review_Influence_Score','Social_Ads_Influence_Score']].melt(var_name='Influence Factor',value_name='Score')
influence_comparison['Influence Factor']=influence_comparison['Influence Factor'].replace({'Review_Influence_Score':'Customer Reviews',
'Social_Ads_Influence_Score':'Social Media Ads'})
sns.countplot(data=influence_comparison,x='Score',hue='Influence Factor',palette=['#1F4E79','#6495B1'])
plt.title('Customer reviews vs Social Media Ads Influence')
plt.xlabel('Influence Score')
plt.ylabel('Number of Respondents')
plt.legend(title='Influence Factor')
plt.tight_layout()
plt.show()
```

Comparison of all the purchase influence factors

```python
influence_cols=[
    'Review_Influence_Score',
    'Social_Ads_Influence_Score',
    'Price_Comparison_Frequency',
    'Discount_Importance_Score',
    'Likely_To_Buy_After_Ad_Score'
]
sns.boxplot(data=df[influence_cols],orient='h',color='lightsteelblue')
plt.title('Comparison of Purchase Influence Factors')
plt.xlabel('Score')
plt.ylabel('Factors')
plt.tight_layout()
plt.show()
```

Insight: Customer reviews and social media advertisements show relatively similar average influence on purchasing decisions, indicating that both contribute to customer buying behaviour. However, price comparison and discount importance have higher average scores than review and social media influence, suggesting that customers place greater emphasis on evaluating prices and finding attractive deals before purchasing. The likelihood of buying after seeing an advertisement is comparatively lower, indicating that advertisements do not always translate into actual purchase interest.

Recommendation: Strengthen customer trust by encouraging genuine customer reviews, ratings and feedback while using targeted social media ads to improve product visibility. Also focus on competitive pricing, transparent discounts and attractive promotional offers, as customers appear more sensitive to price-related factors. Rather than relying only on advertisements, businesses should combine promotions with detailed product information, customer testimonials and value-for-money offers to encourage purchase decisions and improve conversion rates.

5. How frequently do respondents make impulse purchases and experience regret? Are the scores associated? Will they return again?

```python
sns.countplot(data=df,x='Impulse_Purchase_Frequency',hue='Return_Frequency',palette=['#1F4E79','#4682B4','#A7C7E7'])
plt.title('Impulse Purchase Frequency vs Return Frequency')
plt.xlabel('Impulse Purchase Frequency')
plt.ylabel('Number of Respondents')
plt.legend(title='Return Frequency')
plt.tight_layout()
plt.show()
```

```python
regret_impulse=pd.crosstab(df['Impulse_Purchase_Frequency'],df['Regret_After_Purchase_Frequency'])
sns.heatmap(regret_impulse,annot=True,fmt='d',cmap='Blues')
plt.title('Impulse Purchase vs Purchase Regret')
plt.xlabel('Regret After Purchase')
plt.ylabel('Impulse Purchase Frequency')
plt.tight_layout()
plt.show()
```

```python
regret_return=pd.crosstab(df['Regret_After_Purchase_Frequency'],df['Return_Frequency'],normalize='index')*100
sns.heatmap(regret_return,annot=True,fmt='.1f',cmap='Blues')
plt.title('Return Frequency by Purchase Regret (%)')
plt.xlabel('Return Frequency')
plt.ylabel('Purchase Regret Frequency')
plt.tight_layout()
plt.show()
```

Insight: Higher impulse-purchase frequency generally shows an association with higher purchase regret. Customers with lower impulse-purchase frequency tend to report lower regret scores. However, higher purchase regret does not necessarily lead to more frequent product returns. Interestingly, customers with the highest regret score have the highest percentage of respondents who never return products(53.33%). Overall, impulse purchasing, purchase regret and product returns show different behavioural patterns, suggesting that regret does not always translate into return behaviour.

Recommendation: Businesses should encourage informed purchasing decisions by providing accurate product descriptions, clear images, authentic customer reviews and cart-review reminders to reduce impulsive purchases and potential regret. They should also collect post-purchase feedback to understand why dissatisfied customers choose not to return products. Simplifying return procedures, offering transparent return policies and improving post-purchase support may help address customer dissatisfaction and improve the overall shopping experience.

```python
fig,axes=plt.subplots(1,2,figsize=(10,5))
sns.regplot(data=df,x='Social_Ads_Influence_Score',y='Likely_To_Buy_After_Ad_Score',x_jitter=0.12,y_jitter=0.12,ax=axes[0])
axes[0].set_title('Social Advertisement Influence vs Buying Intention')
axes[0].set_xlabel('Social Ads Influence (1–5)')
axes[0].set_ylabel('Likely to Buy After Ad (1–5)')
sns.regplot(data=df,x='Review_Influence_Score',y='Likely_To_Buy_After_Ad_Score',x_jitter=0.12,y_jitter=0.12,ax=axes[1])
axes[1].set_title('Review Influence vs Buying Intention')
axes[1].set_xlabel('Review Influence (1–5)')
axes[1].set_ylabel('Likely to Buy After Ad (1–5)')
plt.tight_layout()
plt.show()
```

Insight: Both Social Ads and Reviews successfully drive buying intent. The Review graph has a steeper slope, indicating reviews are a slightly stronger driver of purchase intent than ads. Data points are widely scattered at every influence level. High influence does not guarantee a sale.

Recommendation: Invest heavily in generating and displaying customer reviews, as they have the highest conversion power. Use Social Ads to drive awareness, but ensure landing pages have great reviews to close the sale. Retarget customers who engage with ads/reviews but show low buying intent(likely due to price or trust issues).

## Conclusion

This exploratory data analysis examined customer purchase behavior across demographics, preferences, influence, satisfaction, and post-purchase patterns. Key findings show that most customers belong to younger age group. Online shopping is driven by convenience and variety, with smartphones as the primary device, clothing as the top category, and digital wallets as the preferred payment method. While customers prioritize price comparisons and discounts, reviews and social ads also influence decisions—though with varying impact. Satisfaction remained similar across shopping preference groups, while trust in reviews varied. Post-purchase analysis revealed a link between impulse buying and regret, but regret did not consistently lead to returns. Overall, look beyond initial purchases to develop customer-focused marketing, competitive pricing, better product info, improved shopping experiences, and stronger post-purchase support to drive long-term engagement.
